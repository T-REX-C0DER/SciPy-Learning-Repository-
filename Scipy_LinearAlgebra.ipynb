```python
import numpy as np 
from scipy import linalg
```

# Linear Algebra

The scipy.linalg() provides fas , stable & feature rich linear algebra routines . This module is build on top of optimized lib .

```python
# The numpy matrix
mat1 = np.array([[1,2,] , [5,6]])
print(mat1)

mat2 = np.array([[56,9,8,4] , [5,6,7,8] , [3,4,5,6] , [45,34,54,56]])
print(mat2)
```

# Basic Matrix Operations

```python
print(f"The inverse of mat is {linalg.inv(mat1)}")
print(f"The determinant of mat is {linalg.det(mat1)}")
print(f"The trace of mat is {np.trace(mat1)}")              # numpy, not linalg
print(f"The vector norm of mat is {linalg.norm(mat1)}")
print(f"The pseudo inverse of mat is {linalg.pinv(mat1)}")
print(f"The kronecker product of mat is {np.kron(mat1, mat2)}")  # numpy + 2 args
```

# Matrix Decomposition
It means breaking the matrix into simpler parts

```python
# LU Decomposition (mat1)
P, L, U = linalg.lu(mat1)
print("LU Decomposition of mat1:")
print(f"P (Permutation):\n{P}")
print(f"L (Lower):\n{L}")
print(f"U (Upper):\n{U}")
print(f"Verification P @ L @ U:\n{P @ L @ U}")  # Should equal mat1
```

```python
# QR Decomposition (mat1)
Q, R = linalg.qr(mat1)
print("\nQR Decomposition of mat1:")
print(f"Q (Orthogonal):\n{Q}")
print(f"R (Upper Triangular):\n{R}")
print(f"Verification Q @ R:\n{Q @ R}")  # Should equal mat1
```

```python
# SVD (mat1)
U, S, Vh = linalg.svd(mat1)
print("\nSVD of mat1:")
print(f"U:\n{U}")
print(f"S (Singular Values):\n{S}")
print(f"Vh:\n{Vh}")
```

```python
# Eigenvalue Decomposition (mat1 only - must be square)
eigenvalues, eigenvectors = linalg.eig(mat1)
print("\nEigenvalue Decomposition of mat1:")
print(f"Eigenvalues:\n{eigenvalues}")
print(f"Eigenvectors:\n{eigenvectors}")
```

```python
# Make mat1 positive definite first
mat1_pd = mat1.T @ mat1  # A^T * A is always positive definite

L = linalg.cholesky(mat1_pd, lower=True)
print("\nCholesky Decomposition of mat1^T @ mat1:")
print(f"L (Lower Triangular):\n{L}")
print(f"Verification L @ L.T:\n{L @ L.T}")  # Should equal mat1_pd
```

# Eigen value problems

```python
print(f"The eigen + values values of mat1 is {linalg.eig(mat1)}")
print(f"The values for symmetric mat is {linalg.eigh(mat1)}")
print(f"only eigen values {linalg.eigvals(mat1)}")
print(linalg.eig_banded)
```

# Matrix Functions

```python
print(f"The mat expo {linalg.expm(mat1)}")
print(f"The mat log {linalg.logm(mat1)}")
print(f"The sq root of mat {linalg.sqrtm(mat1)}")
print(f"mat cosin {linalg.cosm(mat1)}")
print(f"mat sin {linalg.sinm(mat1)}")
```

# Special Matrices

```python
print(linalg.toeplitz(mat1 , mat2))
```
