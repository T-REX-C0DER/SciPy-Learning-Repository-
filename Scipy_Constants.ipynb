# Constants In Scipy

It is a module that provides accurate values of the mathematical constants which is used in science

```python
import numpy as np
from scipy import constants 
```

# Categoris Of Constants

```python
# Mathematical Constants
print(f"The value of pi is {constants.pi}")
print(f"The golden ratio value is {constants.golden_ratio}")
```

```python
# Physical Constants
print(f"The speed of light is {constants.c}")
print(f"The gravitational const is {constants.G}")
print(f"The planks const is {constants.h}")
print(f"The botzmann const is {constants.k}")
print(f"The avagadro number is {constants.N_A}")
print(f"The charge on a electron is {constants.e}")
```

```python
# Electromagnetic Constants
print(f"The permitivity of vacuum is {constants.epsilon_0}")
print(f"The permeability of vaccum is {constants.mu_0}")
```

```python
# Atomic & Quatum Constants
print(f"The reduced planks const is {constants.hbar}")
print(f"The fine structure const is {constants.alpha}")
print(f"The mass of electron is {constants.electron_mass}")
```

```python
# Unit Conversion Constants
meters = 3*constants.mile
print(meters)

seconds = 2*constants.hour
print(seconds)
```

```python
# Astronomical Constants
print(f"The earth - sun dist is {constants.astronomical_unit}")
print(f"The light years val is {constants.light_year}")
print(f"The astronomy dist is {constants.parsec}")
```
