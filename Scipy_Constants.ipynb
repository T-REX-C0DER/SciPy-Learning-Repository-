# Constants In Scipy

It is a module that provides accurate values of the mathematical constants which is used in science

```python
import numpy as np
from scipy import constants 
```

# Categoris Of Constants

```python
# Mathematical Constants
print(f"The value of pi is {constants.pi}")
print(f"The golden ratio value is {constants.golden_ratio}")
print(f"The eulers number is {constants.e}")
```
