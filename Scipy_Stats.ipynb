# Probabili Distributions 

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# For reproducibility
np.random.seed(42)
```

# Discrete Distributions

```python
# Bernouli Distribution
# Single trial with success probability p
p = 0.6

# PMF
x = [0, 1]
pmf = stats.bernoulli.pmf(x, p)

print("Mean:", stats.bernoulli.mean(p))
print("Variance:", stats.bernoulli.var(p))

plt.bar(x, pmf)
plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.show()
```

```python
# Binomial Distribution

n = 10
p = 0.5

x = np.arange(0, n+1)
pmf = stats.binom.pmf(x, n, p)

print("Mean:", stats.binom.mean(n, p))
print("Variance:", stats.binom.var(n, p))

plt.bar(x, pmf)
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()
```

```python
# POisson Distribution

lam = 3

x = np.arange(0, 15)
pmf = stats.poisson.pmf(x, lam)

print("Mean:", stats.poisson.mean(lam))
print("Variance:", stats.poisson.var(lam))

plt.bar(x, pmf)
plt.title("Poisson Distribution")
plt.xlabel("Count")
plt.ylabel("Probability")
plt.show()
```

```python
# Geometric Distribution

p = 0.3

x = np.arange(1, 15)
pmf = stats.geom.pmf(x, p)

print("Mean:", stats.geom.mean(p))
print("Variance:", stats.geom.var(p))

plt.bar(x, pmf)
plt.title("Geometric Distribution")
plt.xlabel("Trials Until First Success")
plt.ylabel("Probability")
plt.show()
```

```python
# Hypergeometric Distribution

M = 20   # population size
n = 7    # success states in population
N = 12   # number of draws

x = np.arange(0, n+1)
pmf = stats.hypergeom.pmf(x, M, n, N)

print("Mean:", stats.hypergeom.mean(M, n, N))
print("Variance:", stats.hypergeom.var(M, n, N))

plt.bar(x, pmf)
plt.title("Hypergeometric Distribution")
plt.xlabel("Successes")
plt.ylabel("Probability")
plt.show()
```

# Continuous Distributions 

```python
# Uniform Distribution

a = 0
b = 10

x = np.linspace(a, b, 100)
pdf = stats.uniform.pdf(x, loc=a, scale=b-a)

print("Mean:", stats.uniform.mean(loc=a, scale=b-a))
print("Variance:", stats.uniform.var(loc=a, scale=b-a))

plt.plot(x, pdf)
plt.title("Uniform Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

```python
# Normal Distribution

mu = 0
sigma = 1

x = np.linspace(-5, 5, 200)
pdf = stats.norm.pdf(x, mu, sigma)

print("Mean:", stats.norm.mean(mu, sigma))
print("Variance:", stats.norm.var(mu, sigma))

plt.plot(x, pdf)
plt.title("Normal Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

```python
# Exponential Distribution

lam = 2
scale = 1/lam

x = np.linspace(0, 5, 200)
pdf = stats.expon.pdf(x, scale=scale)

print("Mean:", stats.expon.mean(scale=scale))
print("Variance:", stats.expon.var(scale=scale))

plt.plot(x, pdf)
plt.title("Exponential Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```
